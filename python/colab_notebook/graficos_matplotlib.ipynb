# **Aula 1 - Conhecendo a biblioteca Matplotlib**


##**Conhecendo os dados**

### Importando os dados

```python
import pandas as pd
```

```python
df = pd.read_csv('/content/canadian_immegration_data.csv')
```

```python
df
```

```python
df.info()
```

###Realizando alguns tratamentos

```python
df.set_index('Country', inplace=True)
```

```python
anos = list(map(str, range(1980, 2014)))
```

```python
anos
```

### Extraindo a série de dados para o Brasil

```python
brasil = df.loc['Brazil', anos]
```

```python
brasil
```

```python
brasil_dict = {'ano': brasil.index.tolist(), 'imigrantes': brasil.values.tolist()}
dados_brasil = pd.DataFrame(brasil_dict)
```

```python
dados_brasil
```

##**Criando o primeiro gráfico com Matplotlib**

> Documentação da biblioteca: https://matplotlib.org/

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
```

### Escolhendo quais anos aparecerem no eixo X com `plt.xticks`

```python
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
```

### Alterando os ticks do eixo Y com `plt.yticks`

```python

plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
plt.yticks([500, 1000, 1500, 2000, 2500, 3000])
```

###Exibindo gráfico com `plt.show()`

```python
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
plt.show()
```

### Modificando o tamanho do gráfico com a função `plt.figure()`

```python
plt.figure(figsize=(8,4))
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
plt.show()
```

##**Adicionando títulos e rótulos com Matplotlib**

### Adicionando título com `plt.title()`





```python
plt.figure(figsize=(8,4))
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
plt.title('Imigração do Brasil para o Canadá')
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
plt.show()
```

### Adicionando rótulos aos eixos X e Y usando `plt.xlabel()` e `plt.ylabel()`


```python
plt.figure(figsize=(8,4))
plt.plot(dados_brasil['ano'], dados_brasil['imigrantes'])
plt.title('Imigração do Brasil para o Canadá')
plt.xlabel('Ano')
plt.ylabel('Número de imigrantes')
plt.xticks(['1980', '1985', '1990', '1995', '2000', '2005', '2010'])
plt.show()
```

## SUBPLOTS

```python
fig, ax = plt.subplots(figsize = (8,4))
#fig = figura criada
#ax = eixos (axis)
ax.plot(dados_brasil['ano'], dados_brasil['imigrantes'], lw=3, color='red')

#lw='' espessura da linha
#marker='o' adiciona pontos corretos do gráficos
ax.set_title('Imigração do Brasil para o Canadá \n1980 a 2013', fontsize=18, loc='left')
ax.set_xlabel('Ano', fontsize=14)
ax.set_ylabel('Número de imigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
#Assim terá uma frquência de 5 em 5 anos
plt.grid(linestyle='--')
plt.show()
```

Trabalhando com dois gráficos

```python
fig, axs = plt.subplots(1, 2, figsize = (15,5))
#axs é o número de gráficos
#plt.subplots(Xnumero de linhas, numero de colunas, tamanho da figura)

axs[0].plot(dados_brasil['ano'], dados_brasil['imigrantes'])
axs[0].set_title('Imigração do Brasil para o Canadá \n1980 a 2013')
axs[0].set_xlabel('Ano')
axs[0].set_ylabel('Número de imigrantes')
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].grid()

axs[1].boxplot(dados_brasil['imigrantes'])
axs[1].set_title('Boxplot da imigração do Brasil para o Canadá \n1980 a 2013')
axs[1].set_xlabel('Brasil')
axs[1].set_ylabel('Número de imigrantes')
axs[1].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[1].grid()

plt.show()
```

```python
fig, axs = plt.subplots(2, 2, figsize = (15,12))
#axs é o número de gráficos
#plt.subplots(Xnumero de linhas, numero de colunas, tamanho da figura)
fig.suptitle('Comparação entre os países')
fig.subplots_adjust(hspace=0.5, wspace=0.3)
#fig.subplots_adjust(espaçamento vertical entre os gráficos, espaçamento horizontal entre os gráficos)

axs[0,0].plot(df.loc['Brazil', anos])
axs[0,0].set_title(' Brazil')


axs[0,1].plot(df.loc['Colombia', anos])
axs[0,1].set_title(' Colombia')

axs[1,0].plot(df.loc['Argentina', anos])
axs[1,0].set_title(' Argentina')

axs[1,1].plot(df.loc['Peru', anos])
axs[1,1].set_title(' Peru')

for ax in axs.flat: ##.flat faz com que percorra todos os subpolots
  ax.xaxis.set_major_locator(plt.MultipleLocator(5))
  ax.grid()

for ax in axs.flat:
  ax.set_xlabel('Ano')
  ax.set_ylabel('Número de imigrantes')

ymin = 0
ymax = 7000

for ax in axs.ravel(): #transformar array bidimensional em unidimensional
    ax.set_ylim(ymin, ymax) #ylim(min, max) coloca um limite entre essas duas variáveis

plt.show()
```
